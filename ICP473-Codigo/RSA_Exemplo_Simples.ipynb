# Exemplo simples de RSA usando p = 17 e q = 11

```python
# Escolha de dois números primos diferentes (secretos)
p = 17
q = 11
```

```python
n = p * q
```

```python
fi_n = (p - 1) * (q - 1)
```

```python
print("primos p = ", p, " e q = ", q)
print("n = ", n, "phi(n) = ", fi_n)
```

```python
e = 7 # escolhemos o e tal que mdc(f(n), e) = 1; 1 < e < fi_n
```

```python
print(f"Chave Publica: e = {e}, n = {n}")
```

Esse é um exemplo simples, ilustrativo.

Serve para demostrar que preciso descobrir um valor d, tal que: d * e = fi_n + 1

Isso deu certo por acaso nesse exemplo!!

Mas não é a forma geral de calcular o inverso modular. O correto é usar o algoritmo extendido de Euclides para calcular  d

```python
#ed = (fi_n + 1)
d = (fi_n + 1)/e #O inverso multiplicativo modular de um número inteiro a módulo m é um inteiro x tal que o produto ax é congruente a 1 em relação ao módulo m (ax ≡ 1 (mod m))
d = int(d) #precisa sempre trabalhar com numeros INTEIROS
```

```python
d = pow(e, -1, fi_n) # Forma correta de calcular -> https://docs.python.org/3/whatsnew/3.8.html
```

```python
print(f"Chave Privada: d = {d}, n = {n}")
```

Se escolher um valor de e que nao seja correto..... O Python vai reclamar porque você escolheu um valor de e que não é coprimo com fi_n.

Matematicamente, o inverso modular só existe se:

mdc(e,fi_n)=1

Se o mdc não for 1, não tem como resolver a equação:

d⋅e≡1(mod fi_n)

pois não há nenhum número d que satisfaça essa congruência.
