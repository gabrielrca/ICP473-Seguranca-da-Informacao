```python
import socket
import ssl
```

```python
# Alvo (pode mudar para qualquer site HTTPS)
host = "ufrj.br"
port = 443
```

```python
# Cria contexto TLS padrão (usa as CAs do sistema)
context = ssl.create_default_context()
```

```python
# Para certificado auto assinado
context.check_hostname = False
context.verify_mode = ssl.CERT_NONE
```

```python
# Abre conexão TCP e "encapsula" em TLS
sock = socket.create_connection((host, port))
print(f"Conectado a: {host}:{port}")
```

```python
ssock = context.wrap_socket(sock, server_hostname=host)
print("-------------------------------")
print("Versão TLS negociada:", ssock.version())
print("Cipher negociado:", ssock.cipher())
```

```python
# Obtém certificado do servidor
cert = ssock.getpeercert()
print("\nInformações do certificado:")
print("Emitido para (Subject):", dict(x[0] for x in cert["subject"]))
print("Emitido por (Issuer):", dict(x[0] for x in cert["issuer"]))
print("Válido de:", cert["notBefore"])
print("Válido até:", cert["notAfter"])
```

```python
ssock.close()
sock.close()
```
