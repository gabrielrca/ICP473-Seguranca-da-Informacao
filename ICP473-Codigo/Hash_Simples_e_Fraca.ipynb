# Hash simples e fraca - Demonstração

```python
# Entrada binária
bin_str = "110011000101010111000111"  # 01011110
#bin_str = "010101011100011111001100"  # 01011110
#bin_str = "110011000101010111000111110011000101010111000111110011000101010111000111"  # 01011110
```

```python
decimal = int(bin_str, 2)
print(decimal)
hexadecimal = hex(decimal)
print(hexadecimal)
```

```python
# Divide em blocos de 8 bits
blocks = []  # lista vazia para guardar os blocos
for i in range(0, len(bin_str), 8):  # vai de 0 até o fim, pulando de 8 em 8
    bloco = bin_str[i:i+8]  # pega os 8 bits atuais
    blocks.append(bloco)   # adiciona à lista de blocos

print("Blocos:", blocks)
```

```python
# Inicializa o resultado como lista de 8 bits
hash_bits = []

# Para cada posição de bit (0 a 7)
for i in range(8):

    xor_bit = 0
    col_bits = []

    # Pega o i-ésimo bit de cada bloco e faz XOR
    for b in blocks:
        bit = int(b[i])
        #print(bit)
        col_bits.append(str(bit))
        xor_bit ^= bit

    hash_bits.append(str(xor_bit))

    print(f"Bits da coluna {i}: {col_bits} -> XOR: {xor_bit}")
```

```python
# Resultado final
hash_result = ''.join(hash_bits)
print("Hash final (8 bits):", hash_result)
```
