```python
frase="the snow lay thick on the steps and the snowflakes driven by the wind looked black in the headlights of the cars"
resultado = ""
letras_vistas = set()

for letra in frase:
    if letra != " " and letra not in letras_vistas:
        resultado += letra
        letras_vistas.add(letra)


print(resultado)
```

```python
alfabeto = "abcdefghijklmnopqrstuvwxyz"
for letra in alfabeto:
    if letra not in letras_vistas:
        resultado += letra

print(resultado)
```
