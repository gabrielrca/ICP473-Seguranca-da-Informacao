```python
primo_q_publico = 353
alpha_publico = 3
```

```python
chave_Xa_secreta = 97
chave_Xb_secreta = 233
```

```python
# A calcula:
chave_Ya_publico = (alpha_publico ** chave_Xa_secreta) % primo_q_publico
```

```python
print(chave_Ya_publico)
```

## B calcula:

```python
chave_Yb_publico = (alpha_publico ** chave_Xb_secreta) % primo_q_publico
```

```python
print(chave_Yb_publico)
```
